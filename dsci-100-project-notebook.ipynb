#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r


library(tidymodels)

# Split the data into training and testing sets
set.seed(1234567)
data_split <- initial_split(tidy_data, prop = 0.75,
                            strata = sweet_sixteen)

training_data <- training(data_split)
testing_data <- testing(data_split)

# Standardize the predictors
knn_recipe <- recipe(sweet_sixteen ~ seed + conf_wins_pct,
                     data = training_data) |>
  step_normalize(all_numeric_predictors())

# Set up the KNN classification model
knn_spec <- nearest_neighbor(
  weight_func = "rectangular",
  neighbors = tune()
) |>
  set_engine("kknn") |>
  set_mode("classification")

# Combine the recipe and model
knn_workflow <- workflow() |>
  add_recipe(knn_recipe) |>
  add_model(knn_spec)

# Create 5 folds for cross-validation
set.seed(1234567)
knn_folds <- vfold_cv(training_data, v = 5,
                      strata = sweet_sixteen)

# Test different values of K
k_values <- tibble(neighbors = c(3, 5, 7, 9, 11, 15, 21))

knn_results <- tune_grid(
  knn_workflow,
  resamples = knn_folds,
  grid = k_values,
  metrics = metric_set(accuracy)
)

# Find the best K
best_k <- select_best(knn_results, metric = "accuracy")
best_k

# Train the final model and test its predictions
final_workflow <- finalize_workflow(knn_workflow, best_k)

final_fit <- last_fit(
  final_workflow,
  split = data_split,
  metrics = metric_set(accuracy, sens)
)

# Display model performance
collect_metrics(final_fit)

# Display the confusion matrix
test_predictions <- collect_predictions(final_fit)

conf_mat(test_predictions,
         truth = sweet_sixteen,
         estimate = .pred_class)

```
